```python
import pandas as pd 
donnes=pd.read_csv("athlete_events.csv")
donnes['Medal']=donnes['Medal'].fillna("None")
donnes_NaN_supprimes=donnes.dropna(how='any')
donnes[['Age','Height','Weight']]=donnes[['Age','Height','Weight']].fillna(donnes[['Age','Height','Weight']].mean())
donnes_NaN_remplaces=donnes.drop_duplicates()
print('tableau avec NaN remplacés')
print(donnes_NaN_remplaces.shape)
print('tableau avec suppression des lignes contenant des NaN')
print(donnes_NaN_supprimes.shape)
```
